```python
import torch
from torch import nn
from torch.nn import functional as F
from d2l import torch as d2l

class Residual(nn.Module):
    """残差块"""
    def __init__(self, input_channels, num_channels, use_1x1conv=False, strides=1):
        super().__init__()
        # 第一个卷积层：3x3，步幅可能为2（用于下采样）
        self.conv1 = nn.Conv2d(input_channels, num_channels,
                               kernel_size=3, padding=1, stride=strides)
        # 第二个卷积层：3x3，步幅为1，保持形状
        self.conv2 = nn.Conv2d(num_channels, num_channels,
                               kernel_size=3, padding=1)
        
        # 批量归一化层
        self.bn1 = nn.BatchNorm2d(num_channels)
        self.bn2 = nn.BatchNorm2d(num_channels)
        
        # 1x1卷积层：用于调整捷径分支的通道数和尺寸（当需要时）
        if use_1x1conv:
            self.conv3 = nn.Conv2d(input_channels, num_channels,
                                   kernel_size=1, stride=strides)
        else:
            self.conv3 = None

    def forward(self, X):
        # 主路径：Conv -> BN -> ReLU -> Conv -> BN
        Y = F.relu(self.bn1(self.conv1(X)))
        Y = self.bn2(self.conv2(Y))
        
        # 捷径分支：如果需要，用1x1卷积调整X的形状
        if self.conv3:
            X = self.conv3(X)
        
        # 残差连接：主路径输出 + 捷径输出，然后经ReLU
        Y += X
        return F.relu(Y)
```

```python
# 测试：输入输出形状一致（恒等映射）
blk = Residual(3, 3)
X = torch.rand(4, 3, 6, 6)
print(blk(X).shape)  # torch.Size([4, 3, 6, 6])

# 测试：翻倍通道数并减半高宽（use_1x1conv=True, strides=2）
blk = Residual(3, 6, use_1x1conv=True, strides=2)
print(blk(X).shape)  # torch.Size([4, 6, 3, 3])
```

```python
b1 = nn.Sequential(
    nn.Conv2d(1, 64, kernel_size=7, stride=2, padding=3),
    nn.BatchNorm2d(64), 
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=3, stride=2, padding=1)
)
# 输入: (1, 224, 224) → 输出: (64, 56, 56)
```

```python
def resnet_block(input_channels, num_channels, num_residuals, first_block=False):
    """构建一个 ResNet 模块，包含多个残差块"""
    blk = []
    for i in range(num_residuals):
        if i == 0 and not first_block:
            # 第一个残差块：需要下采样（通道翻倍，尺寸减半）
            blk.append(Residual(input_channels, num_channels,
                                use_1x1conv=True, strides=2))
        else:
            # 后续残差块：保持形状
            blk.append(Residual(num_channels, num_channels))
    return blk

# 构建完整的 ResNet-18
b2 = nn.Sequential(*resnet_block(64, 64, 2, first_block=True))   # 输出: (64, 56, 56)
b3 = nn.Sequential(*resnet_block(64, 128, 2))                    # 输出: (128, 28, 28)
b4 = nn.Sequential(*resnet_block(128, 256, 2))                   # 输出: (256, 14, 14)
b5 = nn.Sequential(*resnet_block(256, 512, 2))                   # 输出: (512, 7, 7)
```

```python
net = nn.Sequential(
    b1, b2, b3, b4, b5,
    nn.AdaptiveAvgPool2d((1, 1)),  # 全局平均池化 → (512, 1, 1)
    nn.Flatten(),                   # 展平 → 512
    nn.Linear(512, 10)              # 全连接层输出10类（Fashion-MNIST）
)
```

```python
X = torch.rand(size=(1, 1, 224, 224))
for layer in net:
    X = layer(X)
    print(layer.__class__.__name__, 'output shape:\t', X.shape)
```
