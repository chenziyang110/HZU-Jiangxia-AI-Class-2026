# 6.3 填充和步幅

填充用于控制输出大小并保留边缘信息；步幅用于控制卷积窗口移动距离，从而降低输出的高和宽。


```python
import torch
from torch import nn
```

## 计算卷积层输出形状

为了便于观察二维输出，把输入先扩展为四维，再在计算后去掉批量和通道维度。


```python
def comp_conv2d(conv2d, X):
    X = X.reshape((1, 1) + X.shape)
    Y = conv2d(X)
    return Y.reshape(Y.shape[2:])


X = torch.rand(size=(8, 8))
```

## 填充

当卷积核大小为 `3 x 3` 且四周填充 1 个像素时，输出高宽与输入保持一致。


```python
conv2d = nn.Conv2d(1, 1, kernel_size=3, padding=1)
print(comp_conv2d(conv2d, X).shape)
```

卷积核高宽不同时，可以分别设置高度和宽度方向的填充。


```python
conv2d = nn.Conv2d(1, 1, kernel_size=(5, 3), padding=(2, 1))
print(comp_conv2d(conv2d, X).shape)
```

## 步幅

步幅越大，卷积窗口移动得越快，输出尺寸越小。


```python
conv2d = nn.Conv2d(1, 1, kernel_size=3, padding=1, stride=2)
print(comp_conv2d(conv2d, X).shape)

conv2d = nn.Conv2d(1, 1, kernel_size=(3, 5), padding=(0, 1), stride=(3, 4))
print(comp_conv2d(conv2d, X).shape)
```

常用经验：如果希望输出尺寸约为输入的一半，可以把步幅设为 2；如果希望保持尺寸，常给奇数大小卷积核设置对称填充。
