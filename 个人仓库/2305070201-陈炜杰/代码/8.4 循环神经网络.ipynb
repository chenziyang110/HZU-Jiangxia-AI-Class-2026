# 8.4 循环神经网络

循环神经网络（RNN）通过隐状态保存历史信息，使当前时间步的输出不仅依赖当前输入，也依赖之前的序列。


```python
import torch
```

## 隐状态计算


```python
X = torch.normal(0, 1, (3, 1))
W_xh = torch.normal(0, 1, (1, 4))
H = torch.normal(0, 1, (3, 4))
W_hh = torch.normal(0, 1, (4, 4))

print(torch.matmul(X, W_xh) + torch.matmul(H, W_hh))
print(torch.matmul(torch.cat((X, H), 1), torch.cat((W_xh, W_hh), 0)))
```

上面两种写法等价。把输入和隐状态拼接、把权重拼接，可以把两次矩阵乘法合并为一次矩阵乘法。


## 字符级语言模型

在字符级语言模型中，每个字符被视作一个词元。RNN 读取前面的字符，并预测下一个字符的概率分布。


## 困惑度

困惑度是语言模型常用评价指标，可以理解为模型在每个时间步平均面对多少个候选词元。困惑度越低，语言模型越好。
