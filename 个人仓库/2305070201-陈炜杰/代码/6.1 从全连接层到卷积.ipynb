# 6.1 从全连接层到卷积

本节主要理解为什么图像任务不能直接依赖全连接层，以及卷积神经网络利用了图像中的两个先验：平移不变性和局部性。


## 平移不变性和局部性

- 平移不变性：同一个目标出现在图像不同位置时，前面几层应产生相似响应。
- 局部性：前面几层优先关注相邻像素构成的局部模式，而不是一开始就连接整张图像。
- 卷积层通过共享卷积核参数，在整张图像上重复扫描局部窗口。


```python
def full_connection_params(image_h, image_w, hidden_units):
    """估算把图像展平成向量后接一个全连接隐藏层的参数量。"""
    return image_h * image_w * hidden_units


def conv_layer_params(kernel_h, kernel_w, in_channels, out_channels):
    """估算一个二维卷积层的权重参数量，不含偏置。"""
    return kernel_h * kernel_w * in_channels * out_channels


print('1000x1000 图像接 1000 个隐藏单元的全连接参数量:', full_connection_params(1000, 1000, 1000))
print('11x11 卷积核、3 输入通道、96 输出通道的参数量:', conv_layer_params(11, 11, 3, 96))
```

全连接层参数量会随着图像像素数迅速膨胀；卷积层只学习局部窗口上的共享权重，因此参数更少，也更适合保留图像空间结构。
